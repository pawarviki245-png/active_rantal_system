```python
activa_count = 20
rent_price_per_day = 500
secure_code = "admin123"
rentals = []

def print_menu():
    print("\n--- Welcome to Cyber Success Activa Rental House ---")
    print(f"Total Activas Available: {activa_count}")
    print("1. Rent\n2. Return\n3. Owner\n4. Exit")

def rent_activa():
    global activa_count

    if activa_count <= 0:
        print("No Activas available right now.")
        return

    print("\n--- Rent Activa ---")
    print("T&C:\n1. Aadhaar is mandatory\n2. Driving License is mandatory\n3. ₹500 per day per Activa")
    confirm = input("Do you want to proceed? (Y/N): ").strip().upper()
    if confirm != "Y":
        return

    try:
        rent_num = int(input("Enter number of Activas to rent: "))
        if rent_num > activa_count:
            print("Not enough Activas available.")
            return

        name = input("Enter your name: ")
        phone = input("Enter phone number: ")
        address = input("Enter address: ")
        aadhaar = input("Enter Aadhaar number: ")
        license = input("Enter Driving License number: ")
        days = int(input("Enter number of days: "))

        total = rent_num * days * rent_price_per_day
        activa_count -= rent_num

        rentals.append({
            "name": name,
            "phone": phone,
            "address": address,
            "aadhaar": aadhaar,
            "license": license,
            "days": days,
            "count": rent_num,
            "status": "Pending"
        })

        print(f" Booking Successful! Total Bill: ₹{total}")
        # Save bill
        with open(f"{aadhaar}_bill.txt", "w") as f:
            f.write(f"Customer: {name}\nActivas: {rent_num}\nDays: {days}\nTotal: ₹{total}")

        print(f" Bill saved as {aadhaar}_bill.txt")

    except Exception as e:
        print("Error:", e)

def return_activa():
    global activa_count
    print("\n--- Return Activa ---")
    aadhaar = input("Enter your Aadhaar number: ")
    user_found = False

    for rental in rentals:
        if rental['aadhaar'] == aadhaar and rental['status'] == "Pending":
            try:
                count_return = int(input("Enter number of Activas to return: "))
                if count_return > rental['count']:
                    print("Return count cannot be more than rented count.")
                    return
                rental['count'] -= count_return
                activa_count += count_return

                if rental['count'] == 0:
                    rental['status'] = "Returned"
                print("Return successful.")
                user_found = True
                break
            except:
                print("Invalid input.")
                return

    if not user_found:
        print("No matching pending rental found.")

def owner_view():
    print("\n--- Owner Access ---")
    code = input("Enter secure code: ")
    if code != secure_code:
        print("Wrong secure code.")
        return

    print("\n--- Booking Records ---")
    for rental in rentals:
        print(f"""
Name: {rental['name']}
Phone: {rental['phone']}
Address: {rental['address']}
Aadhaar: {rental['aadhaar']}
License: {rental['license']}
Activas Booked: {rental['count']}
Days Booked: {rental['days']}
Status: {rental['status']}
        """)

def main():
    while True:
        print_menu()
        choice = input("Enter your choice: ").strip()

        if choice == '1':
            rent_activa()
        elif choice == '2':
            return_activa()
        elif choice == '3':
            owner_view()
        elif choice == '4':
            print("Thank you for visiting!")
            break
        else:
            print("Invalid choice. Try again.")
if __name__== "__main__":
    main()
```
